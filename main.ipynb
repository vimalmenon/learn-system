```python
from functools import lru_cache
from contextlib import contextmanager
import time

try:
    from tools import measure_time
except ModuleNotFoundError:
    @contextmanager
    def measure_time():
        start = time.perf_counter()
        try:
            yield
        finally:
            elapsed = time.perf_counter() - start
            print(f"Elapsed time: {elapsed:.6f} seconds")


@lru_cache(maxsize=32)
def fibonacci(n: int) -> int:
    if n < 2:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

with measure_time():
    fib = fibonacci(10)
    print(fib)
with measure_time():
    fib = fibonacci(10)
    print(fib)
with measure_time():
    fib = fibonacci(100)
    print(fib)
```

```python
from abc import ABC, abstractmethod
from typing import Self

class Test(ABC):
    @abstractmethod
    def my_method(self) -> str:
        pass

    @classmethod
    def to_cls(cls) -> Self:
        return cls()

class MyTest(Test):
    def my_method(self) -> str:
        return "Hello from MyTest!"


def main():
    my_test = MyTest.to_cls()
    print(my_test.my_method())

main()
```
